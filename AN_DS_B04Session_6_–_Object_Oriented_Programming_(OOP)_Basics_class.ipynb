
# Constructors

A constructor initializes an object.

Python constructor

```
__init__()
```

It is automatically called whenever an object is created.

---

# Constructor Flow

```
Create Object

↓

__init__()

↓

Initialize Variables

↓

Object Ready
```


````python

# Constructor Example
class Students:
  def __init__(self,st_name,st_age,st_grade,st_total):
    self.name=st_name
    self.age=st_age
    self.grade=st_grade
    self.total=st_total

s1=Students("Akil",23,"5th",400)
s2=Students("Malar",45,"9th",470)
print(f"Name={s1.name}\nAge={s1.age},\nGrade={s1.grade}\n total={s1.total}")
print(f"Name={s2.name}\nAge={s2.age},\nGrade={s2.grade}\n total={s2.total}")

````

````python

# Constructor with Parameters

class Student:

    def __init__(self,name,age):

        self.name = name

        self.age = age

student = Student("Alice",21)

print(student.name)

print(student.age)
````

# Instance Variables

Each object has its own copy.

````python
class Student:

    def __init__(self,name):

        self.name=name

s1=Student("John")

s2=Student("Alice")

print(s1.name)

print(s2.name)
````

# Class Variables

Shared among all objects.

```python

    college="ABC College"

    def __init__(self,name):

        self.name=name

s2=Student("Alice")

print(s1.college)

# Instance vs Class Variables

| Instance Variable     | Class Variable        |
| --------------------- | --------------------- |
| Unique to each object | Shared by all objects |
| Uses self             | Defined inside class  |
| Different values      | Same value            |

# Syntax

```python
class Parent:

    pass

class Child(Parent):

    pass
```

````python
class Animal:
  def sound(self):
    print("It makes some sound")

class Dog(Animal):
  pass
# Example
d1=Dog()
d1.sound()
````

````python
# Multiple Methods
class Person:

    def display(self):

        print("Person")

class Student(Person):

    def study(self):

        print("Studying")

student=Student()

student.display()

student.study()
````


# Method Overriding

Child class provides its own implementation.


````python
class Animal:

    def sound(self):

        print("Animal Sound")

class Dog(Animal):

    def sound(self):

        print("Bark")

dog=Dog()

dog.sound()
````

Encapsulation means protecting data.

Instead of allowing direct access,




![Image](https://images.openai.com/static-rsc-4/Jexk35ZglB2A9oId-dtgydX42CRKfsWdDpH6ZeLK96Rk54-y1abYQuh4vuBGx7rjHlPKfDEEEDt4Y74FQA_-_4PiEPuWCerCBKpukWLiGzNSZT5athi-UgmopJAi-8DNfov8TSoqUYw9JN22jj_EA6ADDhWzL5pp_X5OZ5VKEnGNV0CBwded06TvpK-9D8B1?purpose=fullsize)



````python

# Private Variables


class Bank:

    def __init__(self):

        self.__balance=10000

    def deposit(self,amt):
      self.__balance+=amt
      print(f'Current Balance={self.__balance}')

    def get_balance(self):
      print(f'Balance={self.__balance}')

b1=Bank()

print(b1.get_balance())
print(b1.deposit(4000))
````

````python

# Access Using Methods
class Bank:

    def __init__(self):

        self.__balance=10000

    def getBalance(self):

        return self.__balance

bank=Bank()

print(bank.getBalance())
````


# Polymorphism

One interface

Multiple forms

Example

```
Animal

↓

Dog → Bark

Cat → Meow

Cow → Moo
```

````python
# Example
class Dog:

    def sound(self):

        print("Bark")

class Cat:

    def sound(self):

        print("Meow")

ani=[Dog(),Cat()]

for animal in ani:

    animal.sound()
````

#Abstraction:-

Hiding unneccessary implementation details and showing what user wants

````python
from abc import ABC,abstractmethod

class vechile(ABC):
  @abstractmethod
  def start(self):
    pass

class Car(vechile):
  def start(self):
    print("Car started using engine")


c1=Car()
c1.start()
````
